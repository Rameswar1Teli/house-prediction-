```python
from flask import Flask
from flask import request
from flask import render_template
import joblib
```

```python
app=Flask(__name__,static_folder='static')
model=joblib.load('boston_prediction')
@app.route('/')
def home(name=None):
    return render_template('index.html',name=name)
@app.route('/index',methods=['GET','POST'])
def get_values():
    if(request.method=='POST'):
        rm=float(request.form.get('rm'))
        pt=float(request.form.get('pt'))
        lstat=float(request.form.get('lstat'))
        predict_value=model.predict([[rm,pt,lstat]])
        #render result.html instead of index.html
        return render_template('result.html',prediction=predict_value[0],rm=rm,pt=pt,lstat=lstat)
        #if Get request, redirect to home
        return render_template('index.html')
```

```python
if __name__ == "__main__":
    app.run()
```
